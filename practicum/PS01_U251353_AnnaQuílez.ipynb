<font size="+1" color="blue">**Create function "describe_dataframe".** It must create and display a dataframe containing one row per column, and with the following fields: name of the column ("name"), type ("data_type"), number of non-null values ("count_not_null"), and number of distinct elements ("count_distinct").</font>

```python
def describe_dataframe(df):

    # YOUR CODE HERE
    rows= []

    for column in df.columns:
      rows.append({"name": column,
            "data_type": df[column].dtype,
            "count_not_null": df[column].notnull().sum(),
            "count_distinct": df[column].nunique()})

    input_columns = pd.DataFrame(rows)

    return input_columns

input_columns = describe_dataframe(input_dataset)
display(input_columns)
```

```python
def drop_empty_columns(df):

    # YOUR CODE HERE

    columns_to_keep = []

    for column in df.columns:
        if df[column].notnull().sum() != 0:
            columns_to_keep.append(column)

    return df[columns_to_keep].copy()

df1 = drop_empty_columns(input_dataset)

display(describe_dataframe(df1))
```

```python
def determine_modal_value(df, column):
    
    # YOUR CODE HERE
    modal_value = df[column].mode()[0]
    
    return modal_value
```

# 5. Handle null values in number of bathrooms and bedrooms <font color="blue">[30%]</font>

<font size="+1" color="blue">**Include whatever code you need to decide your null-handling strategy for bathrooms and bedrooms**. Cross-tabulating the data is helpful, we have included one example of this.</font>

<font size="+1" color="blue">**Write your strategy for handling null values in bathrooms and bedrooms**. Justify briefly your choices.</font>

**Answer:** the strategy for handling missing values in `bathrooms`, `bathrooms_text`, `bedrooms` and `beds` will be as follows.

For columns `bathrooms` and `bathroom_text`:
* If bathrooms is null but bathrooms_text contains the number of bathrooms, we will use bathrooms_text to fill bathrooms. If bathrooms_text is null but bathrooms has a value, we will generate the corresponding text from bathrooms.

For columns `bedrooms` and `beds`:
* if bedrooms is null but beds is known, we will calculate bedrooms according to the given rule: 1 bed means 0 bedrooms, and 2 or more beds means beds - 1 bedrooms. If beds is missing we use the number of bedrooms. If both are missing, we remove the row because we cannot know the correct values.

<font size="+1" color="blue">**Implement your strategy for handling null values in bathrooms and bedrooms**. </font>

```python
def handle_null_values_bathrooms_bedrooms(df):
    df_clean = df.copy()

    # null bathrooms
    for i in df_clean.index:
        if pd.isnull(df_clean.loc[i, "bathrooms"]):
            text = df_clean.loc[i, "bathrooms_text"]

            if text in ["Half-bath", "Shared half-bath", "Private half-bath"]:
                df_clean.loc[i, "bathrooms"] = 0.5
            elif not pd.isnull(text):
                df_clean.loc[i, "bathrooms"] = float(text.split()[0])

    # null bathrooms text
    for i in df_clean.index:
        if pd.isnull(df_clean.loc[i, "bathrooms_text"]):
            df_clean.loc[i, "bathrooms_text"] = str(df_clean.loc[i, "bathrooms"]) + " baths"

    # null bedrooms
    for i in df_clean.index:
        if pd.isnull(df_clean.loc[i, "bedrooms"]):
            if df_clean.loc[i, "beds"] == 1:
                df_clean.loc[i, "bedrooms"] = 0
            elif df_clean.loc[i, "beds"] >= 2:
                df_clean.loc[i, "bedrooms"] = df_clean.loc[i, "beds"] - 1

    # null beds
    for i in df_clean.index:
        if pd.isnull(df_clean.loc[i, "beds"]):
            df_clean.loc[i, "beds"] = df_clean.loc[i, "bedrooms"]

    # Remove rows where we still do not have enough information
    df_clean = df_clean.dropna(subset=["bathrooms", "bedrooms", "beds"])

    return df_clean


df3 = handle_null_values_bathrooms_bedrooms(df2)

# This should show no NULL values in bathrooms or bedrooms
display(null_value_counts(df3))
```

# 6. Handle null values in other columns <font color="blue">[20%]</font>

<font size="+1" color="blue">**Write your strategy for the remaining columns with null values**. Indicate for each column if you will (a) drop the rows containig null values, (b) replace by a modal value, (c) replace by a zero or empty string. Justify briefly each choice.</font>

**Answer:** the strategy for handling missing values will be as follows.

Columns related to the description of the property:

* Column `description`: replace with an empty string, as an empty description is valid.

Columns related to minimum and maximum number of nights (`*_nights`):

* These columns are minimum_nights, maximum_nights, minimum_minimum_nights, maximum_minimum_nights, minimum_maximum_nights and maximum_maximum_nights. We will drop the rows containing null values. As we can see, there is only 1 null value in each of these columns, therefore, removing these rows will have a minimal impact and we will not lose much data. Moreover, it does not make sense to replace these values with the mode because we would be inventing a value, and putting 0 would be invalid because we cannot stay a minimum of 0 nights.

Columns related to availability, price, and license:

* These columns are price, price_quote_checkin_date, price_quote_checkout_date, price_quote_total_price, price_quote_price_per_night, price_quote_raw, has_availability, estimated_revenue_l365d and license. Drop the rows. We can not use a listing if we don't know if it is available, its price, or if it has a license.


Columns related to reviews:

* For review scores these, columns are review_scores_rating, review_scores_accuracy, review_scores_cleanliness, review_scores_checkin, review_scores_communication, review_scores_location and review_scores_value. We would replace them with the average value of each column. Whereas for the missing values in reviews_per_month, we put a 0 because a listing with no reviews has zero reviews per month. Furthermore, first_review as well as last_review will remain null, given that they are dates and a null value means that the listing has never been reviewed.

<font size="+1" color="blue">**Implement your strategy for handling the remaining null values**. </font>

```python
def handle_remaining_null_values(df):
    df_clean = df.copy()
    
    # YOUR CODE HERE
    # Description 
    df_clean["description"] = df_clean["description"].fillna("")
    
    # Minimum and maximum 
    df_clean = df_clean.dropna(subset=["minimum_nights","maximum_nights","minimum_minimum_nights", "maximum_minimum_nights","minimum_maximum_nights","maximum_maximum_nights"])

    # Availability, price and license 
    df_clean = df_clean.dropna(subset=["has_availability", "price","price_quote_checkin_date","price_quote_checkout_date","price_quote_total_price","price_quote_price_per_night","price_quote_raw","estimated_revenue_l365d","license"])
    # Review scores 
    review_columns = ["review_scores_rating","review_scores_accuracy","review_scores_cleanliness","review_scores_checkin", "review_scores_communication","review_scores_location","review_scores_value"]

    for column in review_columns:
        average_value = df_clean[column].mean()
        df_clean[column] = df_clean[column].fillna(average_value)
        
    # Reviews 
    df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)
    df_clean["first_review"] = df_clean["first_review"].fillna("")
    df_clean["last_review"] = df_clean["last_review"].fillna("")
    
    return df_clean
    
df4 = handle_remaining_null_values(df3)

# This should show no column with a null value
display(null_value_counts(df4))
```
